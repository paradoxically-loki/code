```python
def swap(a, b):
    temp = a
    a = b
    b = temp

    return a, b

swap(3, 4)
```

```python
# Bubble Sort
# Iteratively sending the largest to the last of the array

arr = [3, 2, 4, 1, 5]

def bubble_sort(arr):
    for i in range(len(arr)-1):
        swapped = False
        for j in range(len(arr)-1-i):
            if(arr[j] > arr[j+1]):
                arr[j+1], arr[j] = arr[j], arr[j+1] # no need for swap. Python has this feature.
                swapped = True

        if swapped == False:
            break

    return arr

bubble_sort(arr)
```

```python
# Selection Sort - O(n^2)
# find the minimum in each iteration and put it at its right place

arr = [3, 2, 4, 1, 5]

def selection_sort(arr):
    for i in range(len(arr) - 1):
        minIdx = i
        for j in range(i+1, len(arr)):
            if(arr[j] < arr[minIdx]):
                minIdx = j

        if(minIdx != i):
            arr[i], arr[minIdx] = arr[minIdx], arr[i]

    return arr

selection_sort(arr)
```
