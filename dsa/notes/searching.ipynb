```python
# Searching in a vector
arr = [1, 2, 3, 4, 5]
print(len(arr))
```

```python
# Linear Search - O(n)
def search(arr, key):   
    for i in range(len(arr)):
        if(arr[i] == key):
            return i
        
    return -1

search(arr, 4)
```

```python
# Binary Search
def binary_search(arr, key):
    n = len(arr)
    if n == 0: return -1
    l = 0
    r = n-1

    while(l <= r):
        m = l + (r-l)//2
        if(arr[m] == key):
            return m
        elif(arr[m] < key):
            l = m+1
        else:
            r = m-1

    return -1

binary_search(arr, 4)
```
