# Graphs
## Storing Graphs

### Adjacency Matrix
- A two dim matrix listing all the nodes and mat[i][j] == 1, indicates an edge bw node i and node j.
- If the graph is undirected, the adj matrix should be symmetric.
### Adjacency List
- For each node, we list the nodes it has as edge with.
### Edge List
- Literally a list of edges in a graph.

```python
edges = [(0, 1), (0, 2), (1, 3), (2, 3), (3, 4)]
num_vertices = 5

# Converting an edge list into adj list
def build_adj_list(edge_list):
    adj_list = {}   

    for u,v in edges:
        if u not in adj_list:
            adj_list[u] = []
        if v not in adj_list: 
            adj_list[v] = []

        adj_list[u].append(v)
        adj_list[v].append(u)
    
    return adj_list

# printing an adjacency list
def print_adj_list(adj_list):
    for node in sorted(adj_list.keys()):
        print(f"{node} -> {adj_list[node]}")


# converting an edge list into adj matrix
def build_adj_mat(edge_list, size):
    matrix = [[0]*size for _ in range(size)]

    for u,v in edge_list:
        matrix[u][v] = 1
        matrix[v][u] = 1

    return matrix


# printing an adj matrix
def print_adj_mat(matrix):
    for i in range(len(matrix)):
        for j in range(len(matrix[0])):
            print(matrix[i][j], end=" ")
        print()


graph_list = build_adj_list(edges)
graph_matrix = build_adj_mat(edges, num_vertices)

print_adj_list(graph_list)
print_adj_mat(graph_matrix)
```

```python
# cycle detection in an undirected graph, using bfs

from collections import deque

class Solution:
	def isCycle(self, V, edges):
	    
	    adj_mat = [[] for _ in range(V)]
        for u, v in edges:
            adj_mat[u].append(v)
            adj_mat[v].append(u)
            
        visit = [False]*V
        q = deque()
        
        
        def bfs(start):
            visit[start] = True
            q.append((start, -1))
            
            while q:
                node, parent = q.popleft()
                for neighbor in adj_mat[node]:
                    if not visit[neighbor]:
                        visit[neighbor] = True
                        q.append((neighbor, node))
                    elif neighbor != parent:
                        return True
                        
            return False
               
            
        for i in range(V):
            if not visit[i]:
                if bfs(i): 
                    return True
                
        return False
		    
```

```python
# cycle detection in an undirected graph, using dfs

from collections import deque

class Solution:
	def isCycle(self, V, edges):
	    
	    adj_mat = [[] for _ in range(V)]
        for u, v in edges:
            adj_mat[u].append(v)
            adj_mat[v].append(u)
            
        visited = [False]*V
        
        def dfs(node, parent):
            visited[node] = True
            
            for neighbor in adj_mat[node]:
                if not visited[neighbor]:
                    if dfs(neighbor, node):
                        return True
                        
                elif neighbor != parent:
                        return True
                        
            return False
            
            
        for i in range(V):
            if not visited[i]:
                if dfs(i, -1): 
                    return True
                    
        return False
```

```python
# cycle detection in a directed graph - dfs

class Solution:
    def isCyclic(self, V: int, edges: list[list[int]]) -> bool:
      
        adj = [[] for _ in range(V)]
        for u, v in edges:
            adj[u].append(v)
            # adj[v].append(u)
            
            
        visited = [False]*V
        pathVisited = [False]*V
        
        
        def dfs(node):
            visited[node] = True
            pathVisited[node] = True
            
            for neighbor in adj[node]:
                if not visited[neighbor]:
                    if dfs(neighbor):
                        return True
                        
                elif pathVisited[neighbor]:
                    return True
                    
            pathVisited[node] = False
            return False
            
            
        for i in range(V):
            if not visited[i]:
                if dfs(i):
                    return True
                    
        return False
```

```python
# Kahn's algo for Topo Sorting, can be considered a bfs technique

from collections import deque

class Solution:
    def topoSort(self, V: int, edges: list[list[int]]) -> list[int]:
        
        adj = [[] for _ in range(V)]
        inDegree = [0]*V
        
        for u, v in edges:
            inDegree[v] += 1
            adj[u].append(v)
            
        topo = []
        q = deque()
        
        for idx, val in enumerate(inDegree):
            if val == 0:
                q.append(idx)
                
        while q:
            node = q.popleft()
            topo.append(node)
            
            for neighbor in adj[node]:
                inDegree[neighbor] -= 1
                if inDegree[neighbor] == 0:
                    q.append(neighbor)
                    
        return topo


    # if len(topo) != V -> cycle, this would be the bfs way of detecting cycle in directed graphs
```
