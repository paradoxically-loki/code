```python
# Random number guessing game
import random
rand_num = random.randint(1,20)
print("Generating a random number bw 1 and 20")

for guessTaken in range(1,7):
    print("take a guess")
    guess = int(input())
    if guess > rand_num: 
        print("Guess taken is too high")
    elif guess < rand_num:
        print("Guess taken is too low")
    else:
        break

if guess == rand_num: 
    print(f"Yay, you guessed it correctly in {guessTaken} trials")
else:
    print(f"Nope, I was thinking of {rand_num}")
```

```python
names = ['A', 'B', 'C', 'D']
print(names.index('C')) #.index() is a module and not a function. Modules are "called on" a variable

# Other modules available for lists are .append(), .insert(), .remove(), etc

# append module
names.append('E')
print(names)

#insert module
names.insert(2, 'F')
print(names)

#remove module
names.remove('F')
print(names)

#sort module
names.sort()
print(names)

names.sort(reverse=True)
print(names)
```

```python
alphnum = [['A', 2], ['C',1], ['E', 3], ['G', 6]]
alphnum.sort(key = lambda x: x[0])
print(alphnum)

alphnum.sort(key = lambda x: x[1])
print(alphnum)
```

```python
# Copy and deep copy

import copy
a = [1, 2, 3, 4, 5]
b = copy.copy(a)
print(a)
print(b)
b[1] = 18
print(a)
print(b)

print(a == b)
print(a is b)
```

```python
c = [1, [2, 3], [4, 5], 6, [7, [8,9]]]
d = copy.copy(c)
print(c)
print(d)
d[1] = 18
print(c)
print(d)

print(c == d)
print(c is d)
```

```python
e = [1, 2, 3, 4, 5]
f = copy.deepcopy(e)
print(e)
print(f)
f[1] = 18
print(e)
print(f)

print( e == f)
print( e is f)
```

```python
g = [1, [2, 3], [4, 5], 6, [7, [8,9]]]
h = copy.deepcopy(g)
print(g)
print(h)
h[1] = 18
print(g)
print(h)

print(g == h)
print(g is h)
```

```python
# Items in a dictionary are unordered, they're indexed by keys
dct = {"name": 'Loki', "age": 23, "college": "IITM"}
# dictionaries have keys(), values(), and items() methods.

for v in dct.values():
    print(v)


for k in dct.keys():
    print(k, dct[k])


for i in dct.items():
    print(i)
```
