# Line Integrals



### Arc Length in 1-dim

$$\int_{a}^{b} ds = \int_{a}^{b} \sqrt{1 + f'(x)^2} dx$$

### Arc Length in 2-dim

$$\int_{C}^{} ds = \int_{t = a}^{t = b} \sqrt{\left({\frac{dx}{dt}}\right)^2 +\left({\frac{dy}{dt}}\right)^2 } dt$$

$t$ comes from parametrization of the path of the integral.

### Line Integral

$$\int_{C}^{} F ds  = \int_{a}^{b} F(r(t)) . r'(t) dt$$
