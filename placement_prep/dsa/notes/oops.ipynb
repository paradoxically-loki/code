### Inheritance

### Multiple Inheritance

```python
# Multiple inheritance - Child class inheriting from multiple parents
# Multi-level inheritance - Classes inheriting at more than one level

class Animal():
    def eat(self):
        print("This animal is eating")

    def sleep(self):
        print("This animal is sleeping")

class Prey(Animal):
    def flee(self):
        print("This animal is fleeing")

class Predator(Animal):
    def hunt(self):
        print("This animal is hunting")

class Rabbit(Prey):
    pass

class Hawk(Predator):
    pass

class Fish(Prey, Predator): # Multiple inheritance
    pass


rabbit = Rabbit()
hawk = Hawk()
fish = Fish()

rabbit.flee()
hawk.hunt()
fish.flee()
fish.hunt()
fish.eat()
fish.sleep()
```

### Abstract Classes


```python
# Abstract Class: A class that can't be instantiated. Meant to be subclassed.
#                 They can contain abstract methods, which are declared but have no implementation
#                 benefits of abstract classes:
#                 1. Prevents instantiation of the class itself
#                 2. Required children to use inherited abstract methods

from abc import ABC, abstractmethod # abc - abstract base class

class Vehicle(ABC):
    
    @abstractmethod
    def go(self):
        pass

    @abstractmethod
    def stop(self):
        pass
```

```python
vehicle = Vehicle() # can't instantiate
```

```python
# If a class is inheriting from an abstract class which has abstractmethods, it must implement all those methods

from abc import ABC, abstractmethod
class Vehicle(ABC):

    @abstractmethod
    def go(self):
        pass

    @abstractmethod
    def stop(self):
        pass


class Car(Vehicle):
    def go(self):
        print("You are driving the car")

    def stop(self):
        print("You stop the car")

class Motorcycle(Vehicle):
    def go(self):
        print("You are driving the motorcycle")

    def stop(self):
        print("You stop the motorcycle")

car = Car()
car.go()
car.stop()

motorcycle = Motorcycle()
motorcycle.go()
motorcycle.stop()
```

### Polymorphism

```python
#polymorphism - Greek word for "having many forms or faces"
#two ways to achieve polymorphism
# 1. Inheritance: an object could be treated of the same type as a parent class
# 2. Duck Typing: object must have necessary attributes and modules

from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass
    

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        print(f"The area of this circle is {3.14*self.radius*self.radius}")

class Square(Shape):
    def __init__(self, width):
        self.width = width

    def area(self):
        print(f"The area of this square is {self.width*self.width}")

class Triangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        print(f"The area of this triangle is {0.5*self.width*self.height}")

shapes = [Circle(4), Square(5), Triangle(5, 6)]

for shape in shapes:
    shape.area()
```

```python
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass
    

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14*self.radius**2
        

class Square(Shape):
    def __init__(self, width):
        self.width = width

    def area(self):
        return self.width**2
        

class Triangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return 0.5*self.width*self.height
        

# let's have a totally different class
class Pizza(Circle):
    def __init__(self, topping, radius):
        self.topping = topping
        super().__init__(radius)


shapes = [Circle(4), Square(5), Triangle(5, 6), Pizza("peperroni", 15)] # pizza is a pizza, and a cirle, and also a shape

for shape in shapes:
    print(shape.area())
```

### Duck Typing

```python
# Duck Typing - another way of achieving polymorphism besides inheritance
#               objects must have the minimum necessary attributes/methods
#               if it looks like a duck, and quacks like a duck, it must be a duck


class Animal():
    alive = True

class Dog(Animal):
    def speak(self):
        print("Woof!")

class Cat(Animal):
    def speak(self):
        print("Meow!")

class Car():
    def speak(self):  # is should be a horn(self), but that won't work in the for loop, so we make a speak() module
        print("HONK!")

    alive = False

animals = [Dog(), Cat(), Car()]

for animal in animals:
    animal.speak()
    print(animal.alive)
```

### Aggregation

```python
# Aggregation - represents a relationship where one object(the whole) contains references to one or more independent objects (the parts)

class Library():
    def __init__(self, name):
        self.name = name
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def list_books(self):
        return [f"{book.title} by {book.author}" for book in self.books]

class Book():
    def __init__(self, title, author):
        self.title = title
        self.author = author

book1 = Book("Lying", "Sam Harris")
book2 = Book("Animal Liberation", "Peter Singer")
book3 = Book("A Lover's Discourse", "Ronald Barthes")

library = Library("The New York Public Library")

library.add_book(book1)
library.add_book(book2)
library.add_book(book3)

library.list_books()
```

### Composition

```python
# Aggregation - A relationship where one object contains a reference to another independent object
# Composition - a composed object directly owns its components, which can not exist independently

class Engine:
    def __init__(self, horse_power):
        self.horse_power = horse_power

class Wheel:
    def __init__(self, size):
        self.size = size

class Car:
    def __init__(self, make, model, horse_power, wheel_size):
        self.make = make
        self.model = model
        self.engine = Engine(horse_power) # this is composition
        self.wheels = [Wheel(wheel_size) for wheel in range(4)] # this is composition

    def display_car(self):
        return f"{self.make} {self.model} {self.engine.horse_power} {self.wheels[0].size}(in)"

car = Car(make='Ford', model='Mustang', horse_power='500hp', wheel_size=18)

print(car.display_car())
```

### Nested Classes

```python
# Nested Classes - A class defined within another class, class outer, and class inner

# Benefits: 
# 1. Allows us to logically group classes that are closely related
# 2. Encapsulates private details that aren't relevant outside of the outer class
# 3. Keeps the namespace clean; reduces the possibility of naming conflict

class Company:
    class Employee:
        def __init__(self, name, position):
            self.name = name
            self.position = position
        
        def get_details(self):
            return f"{self.name} {self.position}"

    def __init__(self, name):
        self.name = name
        self.employees = []
    
    def add_employee(self, name, position):
        new_employee = self.Employee(name, position)
        self.employees.append(new_employee)

    def list_employees(self):
        for employee in self.employees:
            print(employee.get_details())


company = Company("Google")

company.add_employee("Sundar", "CEO")
company.add_employee("Jeff", "President")
company.add_employee("ABC", "SDE2")

company.list_employees()

```

### Static Method

```python
# Static Method - a method that belongs to a class rather than any object from that class(instance)
# static methods are usually used for general utility functions

# Instance Method - best for operations on instances of of the class (objects)
# Static Method - best for utility functions that do not need access to class data

class Employee:
    def __init__(self, name, position):
        self.name = name
        self.position = position

    def get_info(self):
        return f"{self.name} {self.position}"

    @staticmethod
    def is_valid_position(position):
        valid_positions = ["Manager", "Cashier", "Cook", "Janitor"]
        return position in valid_positions

employee1 = Employee("Dario", "CEO")
employee2 = Employee("Christopher", "Head Interpretability")

print(Employee.is_valid_position("Cook"))
print(employee1.get_info())
print(employee2.get_info())
```

### Class Methods

```python
# Class Methods = allow operations related to the class itself
# Take (cls) as the first parameter, which represent the class itself

# Instance methods: best for operations on instances of the class (objects)
# Static methods: best for utility functions that do not need access to class data
# Class methods: best for class level data or require access to the class itself

class Student:
    count = 0
    total_gpa = 0

    def __init__(self, name, gpa):
        self.name = name
        self.gpa = gpa
        Student.count += 1 # increase the count by when whenever a student is instantiated
        Student.total_gpa += gpa

    # Instance method
    def get_info(self):
        return f"{self.name} {self.gpa}"

    # Class method
    @classmethod   #decorator
    def get_count(cls):
        return f"The total number of students are {cls.count}"

    @classmethod
    def get_average_gpa(cls):
        if cls.count ==0: return 0
        return f"The average gpa of the class: {cls.total_gpa/cls.count :.2f}"

student1 = Student("A", 8.9)
student2 = Student("B", 9.2)
student3 = Student("C", 7.5)

print(student1.get_info())
print(Student.get_count())
print(Student.get_average_gpa())
```

### @property decorator

```python
# @property = decorator used to define a method as a property (it can be accessed like an attribute)
# benefits: add additional logic when read, write, or delete attributes
# gives us getter, setter, and deleter method

class Rectangle:
    def __init__(self, width, height):
        self._width = width    #these underscores convey to other develpers that this attribute is supposed to private and should be accessed inside the class
        self._height = height

    @property # these two are getter methods
    def width(self):
        return f"{self._width}(cm)"

    @property
    def height(self):
        return f"{self._height}(cm)"

    # setter methods

    @width.setter
    def width(self, new_width):
        if new_width > 0: self._width = new_width
        else: print("Width must be greater than zero")

    @height.setter
    def height(self, new_height):
        if new_height > 0: self._height = new_height
        else: print("Height must be greater than zero")

    # deleter method

    @width.deleter
    def width(self):
        del self._width
        print("Width has been deleted")

    @height.deleter
    def height(self):
        del self._height
        print("height has been deleted")

    

rectangle = Rectangle(3,4)
print(rectangle.width)
print(rectangle.height)

rectangle.width = 2

print(rectangle._width)
print(rectangle._height)

del rectangle.width

# the above used underscores right before the variable name indicate that those attributes are private
```
