```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('/home/paradox/Code/Neural Network/train.csv')
data = np.array(data)
np.random.shuffle(data)
```

```python
X = torch.tensor(data[:, 1:]/255.0, dtype = torch.float32)
Y = torch.tensor(data[:, 0], dtype = torch.long)
```

```python
X_train, Y_train = X[1000:], Y[1000:]
X_dev, Y_dev = X[:1000], Y[:1000]
```

```python
X_train.shape, Y_train.shape
```

```python
X_dev.shape, Y_dev.shape
```

```python
train_loader = DataLoader(TensorDataset(X_train, Y_train), batch_size = 64, shuffle = True)
```

```python
# Model
model = nn.Sequential(
    nn.Linear(784, 10),
    nn.ReLU(),
    nn.Linear(10, 10)
    # no softmax here - crossentropy loss includes it
)
```

```python
# Training

optimizer = optim.SGD(model.parameters(), lr = 0.1)
loss_fn = nn.CrossEntropyLoss()


for epoch in range(51):
    for X_batch, Y_batch in train_loader:
        optimizer.zero_grad()
        loss = loss_fn(model(X_batch), Y_batch)
        loss.backward()
        optimizer.step()

    if epoch % 10 == 0:
        preds = model(X_train).argmax(dim=1)
        acc = (preds == Y_train).float().mean()
        print(f"Epoch {epoch} - Accuracy: {acc:.4f}")
```
